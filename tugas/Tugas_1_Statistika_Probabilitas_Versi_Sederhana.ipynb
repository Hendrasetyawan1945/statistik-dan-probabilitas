# BAGIAN 1: PRAKTIKUM DARI SLIDE PEMBELAJARAN DOSEN

### 1.1 Praktikum Diskrit: Example 3.8 (Pengiriman Laptop)

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from math import comb
from scipy.integrate import quad

# Fungsi untuk menghitung probabilitas hipergeometrik laptop rusak
def laptop_pmf(k, N, K, n):
    return (comb(K, k) * comb(N - K, n - k)) / comb(N, n)

N = 20  # Total laptop
K = 3   # Rusak
n = 2   # Dibeli sekolah

k_vals = np.arange(0, min(K, n) + 1)
pmf_vals = [laptop_pmf(k, N, K, n) for k in k_vals]
cdf_vals = np.cumsum(pmf_vals)

# Plot PDF (PMF) dan CDF
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].stem(k_vals, pmf_vals, basefmt=" ")
ax[0].set_title('Praktikum Slide - PMF Laptop Rusak')
ax[0].set_xlabel('Banyaknya Laptop Rusak')
ax[0].set_ylabel('Probabilitas')
ax[0].set_xticks(k_vals)
ax[0].grid(True, linestyle='--', alpha=0.6)

ax[1].step(k_vals, cdf_vals, where='mid', color='green', label='CDF')
ax[1].set_title('Praktikum Slide - CDF Laptop Rusak')
ax[1].set_xlabel('Banyaknya Laptop Rusak')
ax[1].set_ylabel('Cumulative Probability')
ax[1].set_xticks(k_vals)
ax[1].grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

print("Tabel Probabilitas Praktikum Slide:")
for k, pmf, cdf in zip(k_vals, pmf_vals, cdf_vals):
    print(f"P(X = {k}) = {pmf:.4f} | F({k}) = {cdf:.4f}")
```

### 1.2 Praktikum Kontinu: Example 3.11 (Kesalahan Suhu Lab)
Fungsi PDF: $f(x) = \frac{x^2}{3}$ untuk $-1 < x < 2$.

```python
%matplotlib inline
def pdf_suhu(x):
    if -1 < x < 2:
        return (x**2) / 3
    return 0.0

pdf_suhu_vec = np.vectorize(pdf_suhu)

def cdf_suhu(x):
    if x < -1: return 0.0
    if x >= 2: return 1.0
    res, _ = quad(pdf_suhu, -1, x)
    return res

cdf_suhu_vec = np.vectorize(cdf_suhu)
x_range = np.linspace(-1.5, 2.5, 500)

# Visualisasi PDF & CDF
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].plot(x_range, pdf_suhu_vec(x_range), 'b-', label='PDF')
ax[0].fill_between(x_range, pdf_suhu_vec(x_range), alpha=0.3, color='blue')
ax[0].set_title('Praktikum Slide - PDF Suhu Lab')
ax[0].grid(True, linestyle='--', alpha=0.6)

ax[1].plot(x_range, cdf_suhu_vec(x_range), 'g-', label='CDF')
ax[1].set_title('Praktikum Slide - CDF Suhu Lab')
ax[1].grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

p_0_1, _ = quad(pdf_suhu, 0, 1)
print(f"Probabilitas P(0 < X <= 1) Praktikum Slide = {p_0_1:.6f} (Teori: 1/9 = {1/9:.6f})")
```

# BAGIAN 2: TUGAS MANDIRI DISKRIT - PELEMPARAN 2 KOIN SEIMBANG

### Deskripsi Soal:
- **Eksperimen**: Melempar 2 koin seimbang secara bersamaan.
- **Ruang Sampel**: $S = \{AA, AG, GA, GG\}$, dengan total titik sampel $n(S) = 2^2 = 4$.
- **Variabel Random $X$**: Banyaknya sisi Gambar ($G$) yang muncul $\implies X \in \{0, 1, 2\}$.

#### Perhitungan Manual:
1. **PMF $f(x) = P(X = x)$**:
   - $P(X = 0) = \frac{n(AA)}{4} = \frac{1}{4} = 0.2500$
   - $P(X = 1) = \frac{n(AG, GA)}{4} = \frac{2}{4} = 0.5000$
   - $P(X = 2) = \frac{n(GG)}{4} = \frac{1}{4} = 0.2500$
   - **Total Peluang**: $0.25 + 0.50 + 0.25 = 1.0000$ (Sah)
2. **Peluang Kejadian $P(X \ge 1)$ (Minimal 1 Gambar)**:
   - $P(X \ge 1) = P(X = 1) + P(X = 2) = 0.50 + 0.25 = 0.7500$ (atau $1 - P(X = 0) = 1 - 0.25 = 0.7500$).

```python
%matplotlib inline
# Implementasi Python Kasus Diskrit (2 Koin)
n_koin = 2
p_koin = 0.5
x_diskrit = np.array([0, 1, 2])

# PMF Binomial: C(n, k) * p^k * (1-p)^(n-k)
pmf_koin = np.array([comb(n_koin, k) * (p_koin**k) * ((1 - p_koin)**(n_koin - k)) for k in x_diskrit])
cdf_koin = np.cumsum(pmf_koin)
prob_ge_1 = np.sum(pmf_koin[x_diskrit >= 1])

# Visualisasi PMF & CDF
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].stem(x_diskrit, pmf_koin, basefmt=" ")
ax[0].set_title('PMF: Pelemparan 2 Koin', fontsize=12)
ax[0].set_xlabel('Banyak Sisi Gambar (X)')
ax[0].set_ylabel('Probabilitas f(x)')
ax[0].set_xticks(x_diskrit)
ax[0].set_ylim(0, 0.65)
ax[0].grid(True, linestyle='--', alpha=0.6)

ax[1].step(x_diskrit, cdf_koin, where='mid', color='green', linewidth=2)
ax[1].set_title('CDF: Pelemparan 2 Koin (Fungsi Tangga)', fontsize=12)
ax[1].set_xlabel('Banyak Sisi Gambar (X)')
ax[1].set_ylabel('F(x)')
ax[1].set_xticks(x_diskrit)
ax[1].set_ylim(0, 1.1)
ax[1].grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# Komparasi Manual vs Python
print("="*55)
print(f"{'x':<4} | {'PMF Manual':<12} | {'PMF Python':<12} | {'CDF Python':<12}")
print("="*55)
manual_map = {0: 0.25, 1: 0.50, 2: 0.25}
for x in x_diskrit:
    print(f"{x:<4} | {manual_map[x]:<12.4f} | {pmf_koin[x]:<12.4f} | {cdf_koin[x]:<12.4f}")
print("="*55)
print(f"Total Probabilitas Python       : {np.sum(pmf_koin):.4f} (Valid = 1.0)")
print(f"P(X >= 1) Manual               : 0.7500")
print(f"P(X >= 1) Python               : {prob_ge_1:.4f}")
print(f"Selisih (Manual - Python)       : {abs(0.75 - prob_ge_1):.4f}")
print("-> KESIMPULAN: HASIL MANUAL DAN PYTHON IDENTIK 100%!")
```

# BAGIAN 3: TUGAS MANDIRI KONTINU - DISTRIBUSI UNIFORM f(x) = 1

### Deskripsi Soal:
Variabel random kontinu $X$ menyatakan waktu tunggu kedatangan sinyal paket data (dalam menit) dengan fungsi kepadatan peluang:
$$f(x) = \begin{cases} 1, & 0 \le x \le 1 \\ 0, & \text{di tempat lain} \end{cases}$$

#### Perhitungan Manual:
1. **Syarat 1 (Non-negatif)**: $f(x) = 1 \ge 0$ pada $[0, 1]$ $\implies$ Terpenuhi.
2. **Syarat 2 (Total Luas Area = 1)**: $\int_0^1 1\,dx = [x]_0^1 = 1 - 0 = 1.0000$ $\implies$ Sah sebagai PDF.
3. **Fungsi Kumulatif (CDF)**: $F(x) = \int_0^x 1\,dt = [t]_0^x = x$ untuk $0 \le x \le 1$.
4. **Peluang Interval $P(0.3 \le X \le 0.7)$**:
   $$P(0.3 \le X \le 0.7) = \int_{0.3}^{0.7} 1\,dx = [x]_{0.3}^{0.7} = 0.7 - 0.3 = 0.4000$$

```python
%matplotlib inline
# Implementasi Python Kasus Kontinu f(x) = 1
def pdf_uniform(x):
    if 0 <= x <= 1:
        return 1.0
    return 0.0

pdf_uniform_vec = np.vectorize(pdf_uniform)

def cdf_uniform(x):
    if x < 0: return 0.0
    if x > 1: return 1.0
    res, _ = quad(pdf_uniform, 0, x)
    return res

cdf_uniform_vec = np.vectorize(cdf_uniform)

x_grid = np.linspace(-0.2, 1.2, 500)
pdf_grid = pdf_uniform_vec(x_grid)
cdf_grid = cdf_uniform_vec(x_grid)

# Visualisasi Grafik PDF dan CDF
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Kurva PDF
ax[0].plot(x_grid, pdf_grid, 'b-', linewidth=2, label='f(x) = 1')
x_fill = np.linspace(0.3, 0.7, 100)
ax[0].fill_between(x_fill, pdf_uniform_vec(x_fill), color='orange', alpha=0.5, label='P(0.3 <= X <= 0.7) = 0.40')
ax[0].set_title('PDF: Distribusi Uniform f(x)=1', fontsize=12)
ax[0].set_xlabel('x (Menit)')
ax[0].set_ylabel('f(x)')
ax[0].set_ylim(-0.1, 1.3)
ax[0].grid(True, linestyle='--', alpha=0.6)
ax[0].legend()

# Kurva CDF
ax[1].plot(x_grid, cdf_grid, 'g-', linewidth=2, label='F(x) = x')
ax[1].set_title('CDF: Fungsi Kumulatif F(x)=x', fontsize=12)
ax[1].set_xlabel('x (Menit)')
ax[1].set_ylabel('F(x)')
ax[1].set_ylim(-0.1, 1.3)
ax[1].grid(True, linestyle='--', alpha=0.6)
ax[1].legend()
plt.tight_layout()
plt.show()

# Komparasi Manual vs Python
total_area_py, _ = quad(pdf_uniform, 0, 1)
prob_interval_py, _ = quad(pdf_uniform, 0.3, 0.7)
manual_prob = 0.4000

print("="*55)
print("PEMBUKTIAN INTEGRAL KONTINU (MANUAL VS PYTHON):")
print("="*55)
print(f"Total Area Manual               : 1.0000")
print(f"Total Area Python (quad)        : {total_area_py:.6f}")
print(f"P(0.3 <= X <= 0.7) Manual      : {manual_prob:.6f}")
print(f"P(0.3 <= X <= 0.7) Python      : {prob_interval_py:.6f}")
print(f"Selisih (Manual - Python)       : {abs(manual_prob - prob_interval_py):.2e}")
print("-> KESIMPULAN: HASIL INTEGRAL NUMERIK PYTHON PERSIS SAMA DENGAN HITUNGAN MANUAL!")
```
