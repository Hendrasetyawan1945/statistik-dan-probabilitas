### Instruksi Tugas:
1. Bangkitkan data secara acak dari 2 distribusi kontinu selain Uniform dengan $n > 100$.
2. Hitung statistik deskriptif: **Mean, Modus, Median, Persentil 20, Persentil 80, Varians, Skewness, Kurtosis**.
3. Interpretasikan hasil melalui nilai statistik deskriptif yang diperoleh.

### Dua Kasus yang Dipilih:
1. **Distribusi Eksponensial**: Waktu jeda antar-kedatangan transaksi di payment gateway e-commerce (detik).
2. **Distribusi Lognormal**: Waktu respon / download aset media CDN pada aplikasi mobile (milidetik).

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import pandas as pd

# Set random seed agar hasil konsisten dan reproducible
np.random.seed(42)
n_samples = 500
print(f"Ukuran sampel yang dibangkitkan: n = {n_samples}")
```

## 1. Studi Kasus 1: Distribusi Eksponensial
**Waktu Jeda Antar-Kedatangan Transaksi Payment Gateway E-Commerce (Detik)**
- Laju kedatangan $\lambda = 0.5$ transaksi/detik (skala $\beta = 1/\lambda = 2.0$ detik)

```python
# Pembangkitan data acak eksponensial
scale_exp = 2.0
data_exp = np.random.exponential(scale=scale_exp, size=n_samples)

# Perhitungan 8 Statistik Deskriptif
mean_exp = np.mean(data_exp)
median_exp = np.median(data_exp)
var_exp = np.var(data_exp, ddof=1)
std_exp = np.std(data_exp, ddof=1)
p20_exp = np.percentile(data_exp, 20)
p80_exp = np.percentile(data_exp, 80)
skew_exp = stats.skew(data_exp)
kurt_exp = stats.kurtosis(data_exp)  # excess kurtosis

# Estimasi Modus kontinu menggunakan KDE (Kernel Density Estimation)
kde_exp = stats.gaussian_kde(data_exp)
x_grid_exp = np.linspace(0, np.max(data_exp), 1000)
mode_exp = x_grid_exp[np.argmax(kde_exp(x_grid_exp))]

df_exp = pd.DataFrame({
    'Parameter Statistik': ['Ukuran Sampel (n)', 'Mean', 'Modus (KDE Peak)', 'Median (P50)', 'Persentil 20 (P20)', 'Persentil 80 (P80)', 'Varians Sampel', 'Standar Deviasi', 'Kemencengan (Skewness)', 'Keruncingan (Kurtosis Eksess)'],
    'Nilai Sampel': [n_samples, f"{mean_exp:.4f}", f"{mode_exp:.4f}", f"{median_exp:.4f}", f"{p20_exp:.4f}", f"{p80_exp:.4f}", f"{var_exp:.4f}", f"{std_exp:.4f}", f"{skew_exp:.4f}", f"{kurt_exp:.4f}"],
    'Nilai Teoretis': [n_samples, '2.0000', '0.0000', '1.3863', '0.4463', '3.2189', '4.0000', '2.0000', '2.0000', '6.0000']
})
df_exp
```

```python
# Visualisasi Grafik Kasus 1: Eksponensial
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram & PDF Teoretis
axes[0].hist(data_exp, bins=25, density=True, alpha=0.6, color='#2b5c8f', edgecolor='black', label='Histogram Sampel (n=500)')
x_plot_exp = np.linspace(0, np.max(data_exp), 500)
axes[0].plot(x_plot_exp, stats.expon.pdf(x_plot_exp, scale=scale_exp), 'r-', lw=2.5, label='PDF Teoretis ($\\lambda=0.5$)')
axes[0].axvline(mean_exp, color='green', linestyle='--', lw=2, label=f'Mean ({mean_exp:.2f} s)')
axes[0].axvline(median_exp, color='purple', linestyle=':', lw=2, label=f'Median ({median_exp:.2f} s)')
axes[0].axvline(p20_exp, color='orange', linestyle='-.', lw=1.5, label=f'P20 ({p20_exp:.2f} s)')
axes[0].axvline(p80_exp, color='brown', linestyle='-.', lw=1.5, label=f'P80 ({p80_exp:.2f} s)')
axes[0].set_title('Distribusi Eksponensial: Jeda Waktu Transaksi Payment Gateway', fontsize=11, fontweight='bold')
axes[0].set_xlabel('Waktu Antar-Kedatangan (detik)')
axes[0].set_ylabel('Kepadatan Peluang (Density)')
axes[0].legend(loc='upper right')
axes[0].grid(True, linestyle=':', alpha=0.6)

# Boxplot
axes[1].boxplot(data_exp, vert=False, patch_artist=True, boxprops=dict(facecolor='#a1c4fd', color='#1d4e89'))
axes[1].axvline(p20_exp, color='orange', linestyle='-.', lw=1.5, label=f'P20 ({p20_exp:.2f} s)')
axes[1].axvline(median_exp, color='purple', linestyle=':', lw=2, label=f'Median ({median_exp:.2f} s)')
axes[1].axvline(mean_exp, color='green', linestyle='--', lw=2, label=f'Mean ({mean_exp:.2f} s)')
axes[1].axvline(p80_exp, color='brown', linestyle='-.', lw=1.5, label=f'P80 ({p80_exp:.2f} s)')
axes[1].set_title('Boxplot Sebaran & Deteksi Outlier (Eksponensial)', fontsize=11, fontweight='bold')
axes[1].set_xlabel('Waktu Antar-Kedatangan (detik)')
axes[1].set_yticks([])
axes[1].legend(loc='upper right')
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 2. Studi Kasus 2: Distribusi Lognormal
**Waktu Respon / Download Aset CDN pada Aplikasi Mobile (Milidetik)**
- Parameter: $\mu = 5.7$, $\sigma = 0.5$

```python
# Pembangkitan data acak lognormal
mu_lognorm = 5.7
sigma_lognorm = 0.5
data_lognorm = np.random.lognormal(mean=mu_lognorm, sigma=sigma_lognorm, size=n_samples)

# Perhitungan 8 Statistik Deskriptif
mean_lognorm = np.mean(data_lognorm)
median_lognorm = np.median(data_lognorm)
var_lognorm = np.var(data_lognorm, ddof=1)
std_lognorm = np.std(data_lognorm, ddof=1)
p20_lognorm = np.percentile(data_lognorm, 20)
p80_lognorm = np.percentile(data_lognorm, 80)
skew_lognorm = stats.skew(data_lognorm)
kurt_lognorm = stats.kurtosis(data_lognorm)

kde_lognorm = stats.gaussian_kde(data_lognorm)
x_grid_lognorm = np.linspace(np.min(data_lognorm), np.max(data_lognorm), 1000)
mode_lognorm = x_grid_lognorm[np.argmax(kde_lognorm(x_grid_lognorm))]

df_lognorm = pd.DataFrame({
    'Parameter Statistik': ['Ukuran Sampel (n)', 'Mean', 'Modus (KDE Peak)', 'Median (P50)', 'Persentil 20 (P20)', 'Persentil 80 (P80)', 'Varians Sampel', 'Standar Deviasi', 'Kemencengan (Skewness)', 'Keruncingan (Kurtosis Eksess)'],
    'Nilai Sampel': [n_samples, f"{mean_lognorm:.4f}", f"{mode_lognorm:.4f}", f"{median_lognorm:.4f}", f"{p20_lognorm:.4f}", f"{p80_lognorm:.4f}", f"{var_lognorm:.4f}", f"{std_lognorm:.4f}", f"{skew_lognorm:.4f}", f"{kurt_lognorm:.4f}"],
    'Nilai Teoretis': [n_samples, '338.67', '232.76', '298.87', '196.22', '455.33', '32551.48', '180.42', '1.75', '5.90']
})
df_lognorm
```

```python
# Visualisasi Grafik Kasus 2: Lognormal
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram & PDF Teoretis
axes[0].hist(data_lognorm, bins=25, density=True, alpha=0.6, color='#1e7e34', edgecolor='black', label='Histogram Sampel (n=500)')
x_plot_lognorm = np.linspace(np.min(data_lognorm), np.max(data_lognorm), 500)
axes[0].plot(x_plot_lognorm, stats.lognorm.pdf(x_plot_lognorm, s=sigma_lognorm, scale=np.exp(mu_lognorm)), 'r-', lw=2.5, label='PDF Teoretis ($\\mu=5.7, \\sigma=0.5$)')
axes[0].axvline(mean_lognorm, color='blue', linestyle='--', lw=2, label=f'Mean ({mean_lognorm:.1f} ms)')
axes[0].axvline(median_lognorm, color='purple', linestyle=':', lw=2, label=f'Median ({median_lognorm:.1f} ms)')
axes[0].axvline(p20_lognorm, color='orange', linestyle='-.', lw=1.5, label=f'P20 ({p20_lognorm:.1f} ms)')
axes[0].axvline(p80_lognorm, color='brown', linestyle='-.', lw=1.5, label=f'P80 ({p80_lognorm:.1f} ms)')
axes[0].set_title('Distribusi Lognormal: Waktu Respon Download CDN Aplikasi Mobile', fontsize=11, fontweight='bold')
axes[0].set_xlabel('Latensi Respon (milidetik)')
axes[0].set_ylabel('Kepadatan Peluang (Density)')
axes[0].legend(loc='upper right')
axes[0].grid(True, linestyle=':', alpha=0.6)

# Boxplot
axes[1].boxplot(data_lognorm, vert=False, patch_artist=True, boxprops=dict(facecolor='#b8e994', color='#1e7e34'))
axes[1].axvline(p20_lognorm, color='orange', linestyle='-.', lw=1.5, label=f'P20 ({p20_lognorm:.1f} ms)')
axes[1].axvline(median_lognorm, color='purple', linestyle=':', lw=2, label=f'Median ({median_lognorm:.1f} ms)')
axes[1].axvline(mean_lognorm, color='blue', linestyle='--', lw=2, label=f'Mean ({mean_lognorm:.1f} ms)')
axes[1].axvline(p80_lognorm, color='brown', linestyle='-.', lw=1.5, label=f'P80 ({p80_lognorm:.1f} ms)')
axes[1].set_title('Boxplot Sebaran & Deteksi Outlier (Lognormal)', fontsize=11, fontweight='bold')
axes[1].set_xlabel('Latensi Respon (milidetik)')
axes[1].set_yticks([])
axes[1].legend(loc='upper right')
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
