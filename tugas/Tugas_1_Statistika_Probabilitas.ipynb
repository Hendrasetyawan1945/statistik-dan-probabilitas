# BAGIAN 1: PRAKTIKUM DARI SLIDE PEMBELAJARAN

### 1.1 Praktikum Diskrit: Example 3.8 (Pengiriman Laptop)

```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb
from scipy.integrate import quad

# Fungsi untuk menghitung probabilitas hipergeometrik laptop rusak
def laptop_pmf(k, N, K, n):
    return (comb(K, k) * comb(N - K, n - k)) / comb(N, n)

N = 20  # Total laptop
K = 3   # Rusak
n = 2   # Dibeli sekolah

k_vals = np.arange(0, min(K, n) + 1)
pmf_vals = [laptop_pmf(k, N, K, n) for k in k_vals]

# Plot PDF (PMF)
plt.figure(figsize=(9, 4))
plt.stem(k_vals, pmf_vals, basefmt=" ")
plt.title('Praktikum Slide - PMF Laptop Rusak')
plt.xlabel('Banyaknya Laptop Rusak')
plt.ylabel('Probabilitas')
plt.xticks(k_vals)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# Plot CDF
cdf_vals = np.cumsum(pmf_vals)
plt.figure(figsize=(9, 4))
plt.step(k_vals, cdf_vals, where='mid', color='green', label='CDF')
plt.title('Praktikum Slide - CDF Laptop Rusak')
plt.xlabel('Banyaknya Laptop Rusak')
plt.ylabel('Cumulative Probability')
plt.xticks(k_vals)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

print("Tabel Probabilitas Praktikum Slide:")
for k, pmf, cdf in zip(k_vals, pmf_vals, cdf_vals):
    print(f"P(X = {k}) = {pmf:.4f} | F({k}) = {cdf:.4f}")
```

### 1.2 Praktikum Kontinu: Example 3.11 (Kesalahan Suhu Lab)
Fungsi PDF: $f(x) = \frac{x^2}{3}$ untuk $-1 < x < 2$.

```python
def pdf_suhu(x):
    if -1 < x < 2:
        return (x**2) / 3
    return 0.0

pdf_suhu_vec = np.vectorize(pdf_suhu)

def cdf_suhu(x):
    if x < -1: return 0.0
    if x >= 2: return 1.0
    res, _ = quad(pdf_suhu, -1, x)
    return res

cdf_suhu_vec = np.vectorize(cdf_suhu)

x_range = np.linspace(-1.5, 2.5, 500)

# Visualisasi PDF & CDF
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
ax[0].plot(x_range, pdf_suhu_vec(x_range), 'b-', label='PDF')
ax[0].fill_between(x_range, pdf_suhu_vec(x_range), alpha=0.3, color='blue')
ax[0].set_title('Praktikum Slide - PDF Suhu Lab')
ax[0].grid(True, linestyle='--', alpha=0.6)

ax[1].plot(x_range, cdf_suhu_vec(x_range), 'g-', label='CDF')
ax[1].set_title('Praktikum Slide - CDF Suhu Lab')
ax[1].grid(True, linestyle='--', alpha=0.6)
plt.show()

p_0_1, _ = quad(pdf_suhu, 0, 1)
print(f"Probabilitas P(0 < X <= 1) Praktikum Slide = {p_0_1:.6f} (Teori: 1/9 = {1/9:.6f})")
```

# BAGIAN 2: TUGAS 1 SOAL VARIABEL RANDOM DISKRIT

### Studi Kasus: Pelemparan 3 Koin Seimbang
- **Eksperimen**: Melempar 3 koin seimbang secara serempak.
- **Ruang Sampel ($S$)**: $S = \{AAA, AAG, AGA, GAA, AGG, GAG, GGA, GGG\}$, dengan $n(S) = 2^3 = 8$.
- **Variabel Random $X$**: Banyaknya sisi Gambar ($G$) yang muncul, $X \in \{0, 1, 2, 3\}$.

#### Perhitungan Manual:
- $f(0) = P(X = 0) = 1/8 = 0.125$
- $f(1) = P(X = 1) = 3/8 = 0.375$
- $f(2) = P(X = 2) = 3/8 = 0.375$
- $f(3) = P(X = 3) = 1/8 = 0.125$
- **Total Probabilitas**: $\sum f(x) = 1.000$ (Syarat PMF Terpenuhi)

#### Peluang Kejadian Tertentu:
- Peluang muncul minimal 2 Gambar: $P(X \ge 2) = P(X = 2) + P(X = 3) = \frac{3}{8} + \frac{1}{8} = \frac{4}{8} = 0.500$.

```python
# Implementasi Python untuk Kasus Diskrit (3 Koin)
n_koin = 3
p_koin = 0.5
x_diskrit = np.array([0, 1, 2, 3])

# Menghitung PMF dengan rumus kombinasi binomial: C(n, k) * p^k * (1-p)^(n-k)
pmf_koin = np.array([comb(n_koin, k) * (p_koin**k) * ((1 - p_koin)**(n_koin - k)) for k in x_diskrit])

# Menghitung CDF
cdf_koin = np.cumsum(pmf_koin)

# Menghitung Peluang P(X >= 2)
prob_ge_2 = np.sum(pmf_koin[x_diskrit >= 2])

# Visualisasi Grafik PMF dan CDF
fig, ax = plt.subplots(1, 2, figsize=(14, 4))

# Plot PMF
ax[0].stem(x_diskrit, pmf_koin, basefmt=" ")
ax[0].set_title('PMF: Pelemparan 3 Koin', fontsize=12)
ax[0].set_xlabel('Banyak Sisi Gambar (X)')
ax[0].set_ylabel('Probabilitas f(x)')
ax[0].set_xticks(x_diskrit)
ax[0].set_ylim(-0.05, 0.5)
ax[0].grid(True, linestyle='--', alpha=0.6)

# Plot CDF
ax[1].step(x_diskrit, cdf_koin, where='mid', color='green', linewidth=2)
ax[1].set_title('CDF: Pelemparan 3 Koin', fontsize=12)
ax[1].set_xlabel('Banyak Sisi Gambar (X)')
ax[1].set_ylabel('Cumulative F(x)')
ax[1].set_xticks(x_diskrit)
ax[1].set_ylim(0, 1.1)
ax[1].grid(True, linestyle='--', alpha=0.6)
plt.show()

# Pembuktian Komparasi Manual vs Python
print("="*55)
print(f"{'x':<4} | {'PMF Manual':<12} | {'PMF Python':<12} | {'CDF Python':<12}")
print("="*55)
manual_pmf_map = {0: 1/8, 1: 3/8, 2: 3/8, 3: 1/8}
for x in x_diskrit:
    print(f"{x:<4} | {manual_pmf_map[x]:<12.4f} | {pmf_koin[x]:<12.4f} | {cdf_koin[x]:<12.4f}")
print("="*55)
print(f"Total Probabilitas Python       : {np.sum(pmf_koin):.4f} (Valid = 1.0)")
print(f"P(X >= 2) Manual               : 0.5000")
print(f"P(X >= 2) Python               : {prob_ge_2:.4f}")
print(f"Selisih (Manual - Python)       : {abs(0.5 - prob_ge_2):.4f}")
print("-> KESIMPULAN: HASIL MANUAL DAN PYTHON IDENTIK 100%!")
```

# BAGIAN 3: TUGAS 1 SOAL VARIABEL RANDOM KONTINU

### Studi Kasus: Waktu Respon Sistem Komputasi dengan Fungsi Linier
Variabel random kontinu $X$ menyatakan waktu respon proses (dalam menit) dengan fungsi kepadatan peluang:
$$f(x) = \begin{cases} 2x, & 0 \le x \le 1 \\ 0, & \text{di tempat lain} \end{cases}$$

#### Perhitungan Manual:
1. **Syarat 1**: Untuk $0 \le x \le 1$, nilai $2x \ge 0$ (Non-negatif).
2. **Syarat 2**: $\int_0^1 2x\,dx = [x^2]_0^1 = 1^2 - 0^2 = 1.0$ (Luas Area Total = 1, Sah sebagai PDF).
3. **Fungsi Kumulatif (CDF)**: $F(x) = \int_0^x 2t\,dt = [t^2]_0^x = x^2$ untuk $0 \le x \le 1$.
4. **Peluang Interval $P(0.2 \le X \le 0.6)$**:
   $$P(0.2 \le X \le 0.6) = \int_{0.2}^{0.6} 2x\,dx = [x^2]_{0.2}^{0.6} = 0.6^2 - 0.2^2 = 0.36 - 0.04 = 0.3200$$

```python
# Implementasi Python untuk Kasus Kontinu f(x) = 2x
def pdf_kontinu(x):
    if 0 <= x <= 1:
        return 2 * x
    return 0.0

pdf_kontinu_vec = np.vectorize(pdf_kontinu)

def cdf_kontinu(x):
    if x < 0: return 0.0
    if x > 1: return 1.0
    res, _ = quad(pdf_kontinu, 0, x)
    return res

cdf_kontinu_vec = np.vectorize(cdf_kontinu)

x_grid = np.linspace(-0.2, 1.2, 500)
pdf_grid = pdf_kontinu_vec(x_grid)
cdf_grid = cdf_kontinu_vec(x_grid)

# Visualisasi Grafik PDF dan CDF
fig, ax = plt.subplots(1, 2, figsize=(14, 4))

# Kurva PDF
ax[0].plot(x_grid, pdf_grid, 'b-', linewidth=2, label='f(x) = 2x')
# Warnai area interval 0.2 hingga 0.6
x_fill = np.linspace(0.2, 0.6, 100)
ax[0].fill_between(x_fill, pdf_kontinu_vec(x_fill), color='orange', alpha=0.5, label='P(0.2 <= X <= 0.6) = 0.32')
ax[0].set_title('PDF: Waktu Respon Sistem f(x)=2x', fontsize=12)
ax[0].set_xlabel('x (Menit)')
ax[0].set_ylabel('f(x)')
ax[0].grid(True, linestyle='--', alpha=0.6)
ax[0].legend()

# Kurva CDF
ax[1].plot(x_grid, cdf_grid, 'g-', linewidth=2, label='F(x) = x²')
ax[1].set_title('CDF: Fungsi Kumulatif F(x)=x²', fontsize=12)
ax[1].set_xlabel('x (Menit)')
ax[1].set_ylabel('F(x)')
ax[1].grid(True, linestyle='--', alpha=0.6)
ax[1].legend()
plt.show()

# Pembuktian Komparasi Manual vs Python
total_area_py, _ = quad(pdf_kontinu, 0, 1)
prob_interval_py, _ = quad(pdf_kontinu, 0.2, 0.6)
manual_prob = 0.3200

print("="*55)
print("PEMBUKTIAN INTEGRAL KONTINU (MANUAL VS PYTHON):")
print("="*55)
print(f"Total Area Manual               : 1.0000")
print(f"Total Area Python (quad)        : {total_area_py:.6f}")
print(f"P(0.2 <= X <= 0.6) Manual      : {manual_prob:.6f}")
print(f"P(0.2 <= X <= 0.6) Python      : {prob_interval_py:.6f}")
print(f"Selisih (Manual - Python)       : {abs(manual_prob - prob_interval_py):.2e}")
print("-> KESIMPULAN: HASIL INTEGRAL NUMERIK PYTHON PERSIS SAMA DENGAN HITUNGAN MANUAL!")
```
