```python
# Import library yang dibutuhkan
import numpy as np
import matplotlib.pyplot as plt
from scipy import integrate
import math

print('Library berhasil di-import.')
```

## 1. KASUS VARIABEL RANDOM DISKRIT: PENGAMBILAN SAMPEL SERVER GPU
Sebuah laboratorium memiliki $N=10$ server (terdiri dari $K=3$ server GPU dan 7 server standar). Diambil sampel $n=3$ server tanpa pengembalian.
- Variabel acak $X$: Jumlah server GPU yang terpilih ($X \in \{0, 1, 2, 3\}$).
- Kejadian $A$: Terpilih minimal 2 server GPU ($P(X \ge 2) = P(X=2) + P(X=3)$).

```python
# Parameter kasus diskrit
N = 10  # Total server
K = 3   # Server GPU
n = 3   # Sampel yang diambil

# Fungsi kombinasi
def nCr(n_val, r_val):
    return math.comb(n_val, r_val)

# Total ruang sampel
total_sample_space = nCr(N, n)
print(f'Total kemungkinan kombinasi ruang sampel: {total_sample_space}')

# 1. Perhitungan PMF f(x)
pmf_values = {}
for x in range(4):
    f_x = (nCr(K, x) * nCr(N - K, n - x)) / total_sample_space
    pmf_values[x] = f_x
    print(f'P(X = {x}) = {f_x:.6f}')

print(f'Total peluang sum(f(x)) = {sum(pmf_values.values()):.6f}')

# 2. Perhitungan CDF F(x)
cdf_values = {}
cum_prob = 0.0
for x in range(4):
    cum_prob += pmf_values[x]
    cdf_values[x] = cum_prob
    print(f'F({x}) = P(X <= {x}) = {cdf_values[x]:.6f}')

# 3. Peluang Kejadian A (X >= 2)
prob_A_diskrit = pmf_values[2] + pmf_values[3]
print(f'Peluang Kejadian A (X >= 2): {prob_A_diskrit:.6f}')
```

```python
# Visualisasi Grafik Kasus Diskrit (PMF dan CDF)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Grafik PMF
x_pts = list(pmf_values.keys())
y_pmf = list(pmf_values.values())
ax1.stem(x_pts, y_pmf, linefmt='b-', markerfmt='bo', basefmt='r-')
ax1.set_title('Probability Mass Function (PMF)', fontsize=12, fontweight='bold')
ax1.set_xlabel('Nilai x (Jumlah Server GPU)', fontsize=10)
ax1.set_ylabel('P(X = x)', fontsize=10)
ax1.set_xticks(x_pts)
ax1.grid(True, linestyle='--', alpha=0.6)
for x, y in zip(x_pts, y_pmf):
    ax1.annotate(f'{y:.4f}', (x, y), textcoords='offset points', xytext=(0, 6), ha='center')

# Grafik CDF
x_steps = [-0.5, 0, 1, 2, 3, 3.5]
y_steps = [0, cdf_values[0], cdf_values[1], cdf_values[2], cdf_values[3], 1.0]
ax2.step(x_steps, y_steps, where='post', color='darkgreen', linewidth=2.5)
ax2.plot(list(cdf_values.keys()), list(cdf_values.values()), 'o', color='darkgreen')
ax2.set_title('Cumulative Distribution Function (CDF)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Nilai x', fontsize=10)
ax2.set_ylabel('F(x) = P(X <= x)', fontsize=10)
ax2.set_xticks(range(4))
ax2.set_ylim(-0.05, 1.1)
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 2. KASUS VARIABEL RANDOM KONTINU: WAKTU EKSEKUSI QUERY DATABASE
Waktu eksekusi query dimodelkan sebagai variabel kontinu $X \in [0, 2]$ dengan PDF $f(x) = \frac{3}{8}x^2$.
- Kejadian $A$: Query selesai dalam rentang waktu antara $0.5$ sampai $1.5$ detik ($P(0.5 \le X \le 1.5)$).
- Titik Tunggal: Menunjukkan bahwa $P(X = 1.0) = 0$.

```python
# Definisi fungsi PDF
def f_kontinu(x):
    return (3/8) * (x**2)

# 1. Pembuktian Keabsahan PDF (Total Area = 1)
total_area, _ = integrate.quad(f_kontinu, 0, 2)
print(f'Total Area Kurva PDF (0 sampai 2): {total_area:.6f}')

# 2. Peluang Kejadian A (0.5 <= X <= 1.5)
prob_A_kontinu, _ = integrate.quad(f_kontinu, 0.5, 1.5)
print(f'P(0.5 <= X <= 1.5) = {prob_A_kontinu:.6f}')

# 3. Perhitungan Nilai CDF F(x) = x^3 / 8
cdf_05, _ = integrate.quad(f_kontinu, 0, 0.5)
cdf_15, _ = integrate.quad(f_kontinu, 0, 1.5)
print(f'F(0.5) = {cdf_05:.6f}')
print(f'F(1.5) = {cdf_15:.6f}')
print(f'F(1.5) - F(0.5) = {(cdf_15 - cdf_05):.6f}')

# 4. Pembuktian Peluang Titik Tunggal P(X = 1.0)
prob_point, _ = integrate.quad(f_kontinu, 1.0, 1.0)
print(f'P(X = 1.0) = {prob_point:.6f}')
```

```python
# Visualisasi Grafik Kasus Kontinu (PDF dan CDF)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Grafik PDF
x_vals = np.linspace(0, 2, 200)
y_pdf = f_kontinu(x_vals)
ax1.plot(x_vals, y_pdf, 'b-', linewidth=2.5, label='f(x) = 3/8 x^2')
# Arsir area kejadian A (0.5 <= X <= 1.5)
x_fill = np.linspace(0.5, 1.5, 100)
y_fill = f_kontinu(x_fill)
ax1.fill_between(x_fill, y_fill, color='orange', alpha=0.5, label=f'Kejadian A: P(0.5 <= X <= 1.5) = {prob_A_kontinu:.4f}')
ax1.set_title('Probability Density Function (PDF)', fontsize=12, fontweight='bold')
ax1.set_xlabel('Waktu Eksekusi x (detik)', fontsize=10)
ax1.set_ylabel('f(x)', fontsize=10)
ax1.legend(loc='upper left')
ax1.grid(True, linestyle='--', alpha=0.6)

# Grafik CDF
y_cdf = (x_vals**3) / 8.0
ax2.plot(x_vals, y_cdf, 'r-', linewidth=2.5, label='F(x) = x^3 / 8')
ax2.axvline(0.5, color='gray', linestyle=':', label='x = 0.5 (F=0.0156)')
ax2.axvline(1.5, color='gray', linestyle='--', label='x = 1.5 (F=0.4219)')
ax2.set_title('Cumulative Distribution Function (CDF)', fontsize=12, fontweight='bold')
ax2.set_xlabel('Waktu Eksekusi x (detik)', fontsize=10)
ax2.set_ylabel('F(x) = P(X <= x)', fontsize=10)
ax2.legend(loc='upper left')
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
