## Bagian 1: Variabel Random Diskrit (Praktikum 1 - 3)
### Studi Kasus: Pengiriman Laptop Rusak (Example 3.8)
Sebuah pengiriman berisi 20 laptop serupa, di mana 3 di antaranya rusak. Sekolah secara acak membeli 2 laptop.
Variabel acak $X$ didefinisikan sebagai banyaknya laptop rusak yang terbeli ($X \in \{0, 1, 2\}$).

```python
# 1. Import Library yang dibutuhkan
import numpy as np
import matplotlib.pyplot as plt
from math import comb

# 2. Fungsi untuk menghitung probabilitas hipergeometrik
def laptop_pmf(k, N, K, n):
    """
    Menghitung probabilitas P(X = k)
    k : jumlah item cacat dalam sampel
    N : total item (populasi)
    K : total item cacat dalam populasi
    n : ukuran sampel yang diambil
    """
    return (comb(K, k) * comb(N - K, n - k)) / comb(N, n)

# Parameter distribusi
N = 20  # Total laptop
K = 3   # Laptop cacat
n = 2   # Laptop yang dibeli sekolah

# Nilai-nilai k yang mungkin (0, 1, atau 2)
k_vals = np.arange(0, min(K, n) + 1)

# Menghitung PMF untuk setiap nilai k
pmf_vals = [laptop_pmf(k, N, K, n) for k in k_vals]

# Tampilkan hasil dalam bentuk tabel probabilitas
print("=== TABEL DISTRIBUSI PROBABILITAS (PMF) ===")
for k, pmf in zip(k_vals, pmf_vals):
    print(f"P(X = {k}) = {pmf:.5f}")
```

```python
# 3. Visualisasi Probability Mass Function (PMF)
plt.figure(figsize=(10, 5))
plt.stem(k_vals, pmf_vals, basefmt=" ")
plt.title('Probability Distribution Function (PMF / PDF Diskrit)', fontsize=14)
plt.xlabel('Banyaknya Laptop Rusak (X)', fontsize=12)
plt.ylabel('Probabilitas', fontsize=12)
plt.xticks(k_vals)
plt.ylim(-0.05, 0.8)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# 4. Visualisasi Cumulative Distribution Function (CDF)
cdf_vals = np.cumsum(pmf_vals)

plt.figure(figsize=(10, 5))
plt.step(k_vals, cdf_vals, where='mid', label='CDF', color='green', linewidth=2)
plt.title('Cumulative Distribution Function (CDF)', fontsize=14)
plt.xlabel('Banyaknya Laptop Rusak (X)', fontsize=12)
plt.ylabel('Cumulative Probability F(x)', fontsize=12)
plt.xticks(k_vals)
plt.ylim(0, 1.05)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()

print("=== NILAI CDF F(x) ===")
for k, cdf in zip(k_vals, cdf_vals):
    print(f"F({k}) = P(X <= {k}) = {cdf:.5f}")
```

## Bagian 2: Variabel Random Kontinu (Praktikum 4 - 7)
### Studi Kasus: Kesalahan Suhu Reaksi Kimia (Example 3.11)
Fungsi kepadatan probabilitas (PDF):
$$f(x) = \begin{cases} \frac{x^2}{3}, & -1 < x < 2 \\ 0, & \text{di tempat lain} \end{cases}$$
Akan dihitung probabilitas interval $P(0 < X \le 1)$.

```python
# 1. Import Library untuk integrasi numerik
from scipy.integrate import quad

# 2. Definisikan fungsi PDF f(x)
def pdf(x):
    if -1 < x < 2:
        return (x**2) / 3
    else:
        return 0

# Vektorisasi PDF agar dapat menerima array numpy
pdf_vec = np.vectorize(pdf)

# 3. Definisikan fungsi CDF F(x) via integrasi numerik
def cdf(x):
    if x < -1:
        return 0.0
    elif x >= 2:
        return 1.0
    result, _ = quad(pdf, -1, x)
    return result

cdf_vec = np.vectorize(cdf)

x_vals = np.linspace(-1.5, 2.5, 500)
pdf_vals = pdf_vec(x_vals)
cdf_vals = cdf_vec(x_vals)
```

```python
# 4. Visualisasi Kurva PDF
plt.figure(figsize=(10, 5))
plt.plot(x_vals, pdf_vals, label='PDF f(x) = x²/3', color='blue', linewidth=2)
plt.fill_between(x_vals, pdf_vals, alpha=0.3, color='blue')
plt.title('Probability Density Function (PDF)', fontsize=14)
plt.xlabel('x (Kesalahan Suhu dalam °C)', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
# 5. Visualisasi Kurva CDF
plt.figure(figsize=(10, 5))
plt.plot(x_vals, cdf_vals, label='CDF F(x) = (x³ + 1)/9', color='green', linewidth=2)
plt.title('Cumulative Distribution Function (CDF)', fontsize=14)
plt.xlabel('x (Kesalahan Suhu dalam °C)', fontsize=12)
plt.ylabel('F(x)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```

```python
# 6. Menghitung Nilai Probabilitas P(0 < X <= 1)
prob_0_to_1, _ = quad(pdf, 0, 1)
teori = 1 / 9

print(f"Hasil Komputasi quad(pdf, 0, 1) = {prob_0_to_1:.8f}")
print(f"Hasil Perhitungan Teori (1/9)    = {teori:.8f}")
print(f"Selisih (Error Komputasi)       = {abs(prob_0_to_1 - teori):.2e}")
print("Kesimpulan: Hasil perhitungan numerik Python identik dengan perhitungan manual!")
```
